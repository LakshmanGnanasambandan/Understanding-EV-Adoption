```python
df["Annual_Income_USD"].describe()
```

```python
df["Annual_Income_USD"].isnull().sum()
```

```python
df["Income_Group"] = pd.qcut(
    df["Annual_Income_USD"],
    q=4,
    labels=["Low", "Middle", "Upper Middle", "High"]
)
```

```python
pd.crosstab(
    df["Income_Group"],
    df["Will_Buy_EV"],
    normalize="index"
) * 100
```

```python
bins = [20, 30, 40, 50, 60, 70]
labels = ["20-29", "30-39", "40-49", "50-59", "60-69"]

df["Age_Group"] = pd.cut(df["Age"], bins=bins, labels=labels, right=False)
```

```python
pd.crosstab(
    df["Age_Group"],
    df["Will_Buy_EV"],
    normalize="index"
) * 100
```
