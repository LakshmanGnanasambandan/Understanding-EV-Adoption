```python
bins = [0, 10, 20, 30, 40, 50, float("inf")]
labels = ["0-9", "10-19", "20-29", "30-39", "40-49", "50+"]

df["Commute_Group"] = pd.cut(
    df["Daily_Commute_km"],
    bins=bins,
    labels=labels,
    right=False
)

pd.crosstab(
    df["Commute_Group"],
    df["Will_Buy_EV"],
    normalize="index"
) * 100
```

### Observation

EV purchase intention is highest among respondents with daily commutes of 10–19 km (19.51%) and lowest among those commuting 50+ km (16.36%). ### Observation

Range anxiety shows a strong relationship with EV purchase intention. Respondents with Low range anxiety have the highest purchase intention (19.87%), compared with 6.43% among those with Medium range anxiety and only 0.60% among those with High range anxiety. This suggests that reducing consumers' concerns about driving range could be an important consideration when encouraging EV adoption.

### Observation

Environmental concern shows a strong positive relationship with EV purchase intention. Purchase intention increases consistently from 2.14% among respondents with the lowest environmental concern (Level 1) to 41.40% among respondents with the highest environmental concern (Level 5). This suggests that environmentally conscious consumers represent a potentially important segment for EV adoption.

# EDA Summary

## Key Findings

### 1. Environmental Concern
Environmental concern shows a strong positive relationship with EV purchase intention. Purchase intention increases from 2.14% at Level 1 to 41.40% at Level 5.

### 2. Range Anxiety
Range anxiety shows a strong negative relationship with EV purchase intention. Purchase intention decreases from 19.87% for Low range anxiety to only 0.60% for High range anxiety.

### 3. Subsidy Availability
Respondents with a subsidy available have substantially higher EV purchase intention (27.45%) compared with respondents without a subsidy (2.53%).

### 4. Income
EV purchase intention increases across income groups, from 9.53% among Low-income respondents to 26.22% among High-income respondents.

### 5. Other Variables
Gender, current car type, city type, number of cars owned, commute distance, home charging availability, and workplace charging availability show relatively smaller differences in EV purchase intention.

## Overall Conclusion

The strongest patterns in the descriptive analysis are associated with environmental concern, range anxiety, subsidy availability, and income. These variables should receive greater attention in subsequent analysis and business recommendations.
